## Type Error, Type Checking, and Type Conversion

```python
#The following returns an error because the len() function works with string datatypes, not numbers. Returning a Type Error
len(123)
```

```python
#Use the type() function to return the data type of our object

integer_var = 123
print(f"{integer_var} | Data type:  {type(integer_var)}")

string_var = "Hi there!"
print(f"{string_var} | Data type:  {type(string_var)}")

bool_var = True
print(f"{bool_var} | Data type: {type(bool_var)}")

float_var = 56.0
print(f"{float_var} | Data type:  {type(float_var)}")
```

```python
#Type conversion using the str keyword

#Works because we convert the integer (123) to a string, and can then use len() on the string version of our value
print(len(str(123))) 

#Works because we convert our strings to integers, and can then use addition between the two values, whereas before it would have concatenated them
print(int("123") + int("456"))
```

```python
#Can't do a type conversion like below because it would not make logical sense (can't do addition between abc as an integer and 456 as an integer)
print(int("abc") + int("456"))
```

```python
#Type error mini-project (fix the error)
#original code: print("Number of letters in your name: " + len(input("Enter your name")))

print("Number of letters in your name: " + str(len(input("Enter your name"))))
```

## Mathematical Operations

```python
print("My age: " + str(12)) #String concatenation

print(123 + 456) #Addition

print(7 - 3) #Subtraction

print(3 * 2) #Multiplication
```

```python
print("\n| Division |")
print(10 / 2) #When dividing, we always end up with a FLOAT (implicit typecasting)
print(10 / 3) #When dividing, we always end up with a FLOAT (implicit typecasting)

print(10 // 3) #Integer division (BE CAREFUL, when using this we lose out on decimal precisions and it rounds value down)
```

```python
#Exponentiation operator (10 * 10 * 10)
print(10 ** 3)
```

## Number Manipulation

```python
bmi = 84 / 1.65 ** 2
print(bmi)
```

```python
#Flooring (rounding down and cutting off decimal points)
print(int(bmi))
```

```python
#Round up or down depending on decimal place
print(round(bmi))
print(round(bmi, 2))
```

```python
#Assignment operator
score = 0
print(score)

#User scores a point (adds our value with the += shorthand | Can also use -=, *=, /=)
score += 100
print(score)
```

```python
#F-strings  (Allows us to combine strings with variable values and output all in the same line with minimal typecasting)
print(f"Your score is {score}")
```

## Day 2 Project | Tipping Calculator

```python
print("Welcome to the tip calculator!\n")

bill_total = float(input("What was the bill total? "))
tip_percentage = int(input("Choose a gratuity percentage (0, 10, 12, 15) ")) / 100
number_of_people = int(input("How many people are splitting the bill? "))
tip_amount = bill_total * tip_percentage

split_amount = (bill_total + tip_amount) / number_of_people

print(f"Each person should pay: ${split_amount}")
```
