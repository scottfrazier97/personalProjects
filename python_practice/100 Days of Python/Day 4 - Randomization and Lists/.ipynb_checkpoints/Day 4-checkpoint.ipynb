## Randomization & Lists

```python
import random
```

```python
#help(random)
```

```python
#Return random integer in range [a, b], including both end points. Square brackets denote inclusivity on both numbers. 
random.randint(1, 10)
```

```python
rand_num = random.randint(1, 5)

number_prompt = int(input("Pick a number between 1 and 5: "))

print(f"Computer chose: {rand_num} | You picked: {number_prompt}\n")
if rand_num == number_prompt:
    print("Great job!")
else:
    print("Incorrect! Try again")
```

```python
#Using a custom module from external file
import custom_module

rand_100 = random.randint(1, 100)
print(f"{rand_100} * 7 = {custom_module.my_favorite_number * rand_100}")
```

```python
#Random FLOATING point number
#random() -> x in the interval [0, 1), greater than 0 but up to and not including 1

random_float = round(random.random() * 10, 2)

print(f"{random_float}")
if random_float > 3.14:
    print("Greater than pie")
else:
    print("Not as great as pie")
```

```python
#Random FLOATING point number
#Get a random number in the range [a, b) or [a, b] depending on rounding.

random_float_2 = round(random.uniform(1, 10), 2)

print(f"{random_float_2}")
if random_float_2 > 3.14:
    print("Greater than pie")
else:
    print("Not as great as pie")
```

## Mini-Project | Heads or Tails

```python
overall_counter = 0
heads_counter = 0
tails_counter = 0

while overall_counter < 1000:
    coin_flip = random.randint(1, 2)

    overall_counter += 1

    if coin_flip == 1:
        heads_counter += 1
    else:
        tails_counter += 1

print(f"""
Results:
Heads: |{heads_counter}| {round((heads_counter / 1000) * 100, 2)}%
Tails: |{tails_counter}| {round((tails_counter / 1000) * 100, 2)}%
""")
```
