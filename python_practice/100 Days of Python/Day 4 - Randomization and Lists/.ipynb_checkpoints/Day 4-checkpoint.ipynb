```python
#Random FLOATING point number
#random() -> x in the interval [0, 1), greater than 0 but up to and not including 1

random_float = round(random.random() * 10, 2)

print(f"{random_float}")
if random_float > 3.14:
    print("Greater than pie")
else:
    print("Not as great as pie")
```

```python
#Random FLOATING point number
#Get a random number in the range [a, b) or [a, b] depending on rounding.

random_float_2 = round(random.uniform(1, 10), 2)

print(f"{random_float_2}")
if random_float_2 > 3.14:
    print("Greater than pie")
else:
    print("Not as great as pie")
```

## Mini-Project | Heads or Tails

```python
overall_counter = 0
heads_counter = 0
tails_counter = 0

while overall_counter < 1000:
    coin_flip = random.randint(1, 2)

    overall_counter += 1

    if coin_flip == 1:
        heads_counter += 1
    else:
        tails_counter += 1

print(f"""
Results:
Heads: |{heads_counter}| {round((heads_counter / overall_counter) * 100, 2)}%
Tails: |{tails_counter}| {round((tails_counter / overall_counter) * 100, 2)}%
""")
```

## Lists

* Store data using: [ ]
* Ordered → Items maintain their insertion order.
* Changeable (Mutable) → You can modify, add, or remove items after creation.
* Allow duplicates → Can contain multiple items with the same value.
* Indexable → Items can be accessed by index (positive and negative indexing).
* Heterogeneous → Can store elements of different data types in the same list.
* Supports slicing → You can retrieve parts of the list using slice notation.

```python
#States in order of date they joined the Union
#Order preservation of data structures like lists is important in this case
states_by_join_date = [
    "Delaware", "Pennsylvania", "New Jersey", "Georgia", "Connecticut", "Massachusetts", "Maryland", "South Carolina", "New Hampshire", "Virginia",
    "New York", "North Carolina", "Rhode Island", "Vermont", "Kentucky", "Tennessee", "Ohio", "Louisiana", "Indiana", "Mississippi",
    "Illinois", "Alabama", "Maine", "Missouri", "Arkansas", "Michigan", "Florida", "Texas", "Iowa", "Wisconsin",
    "California", "Minnesota", "Oregon", "Kansas", "West Virginia", "Nevada", "Nebraska", "Colorado", "North Dakota", "South Dakota",
    "Montana", "Washington", "Idaho", "Wyoming", "Utah", "Oklahoma", "New Mexico", "Arizona", "Alaska", "Hawaii"
]
```

```python
#For the enumeration function, we provide idx as the index number of our value in the states list as well as the state argument 
#to track and print the actual value. We also specify 1 as the starting point.

for idx, state in enumerate(states_by_join_date, 1):

    #Example of indexing
    first_letter = state[0]
    
    print(f"{idx} | {state} | {first_letter}")
```

Think about indexing starting at 0 as a means of where each item is in reference from beginning...0 being the starting point enables us to think of the rest of our items like this "our second item in the list being index 1, is 1 position away from beginning, and item 2 is 2 positions away from beginning, etc."

```python
states_copy = states_by_join_date.copy()

#Can change specific values by indexing
states_copy[1] = "Pencilvania"

print(states_by_join_date[1])
print(states_copy[1])
```

```python
#New items get added to the end of an existing list using append()
states_copy.append("Scottland")

#grabbing last item using -1 indexing because our new item "Scottland" was appended to the end
print(states_copy[-1])
```

```python
#Can add many items from another list using extend()
states_copy_extended = states_by_join_date.copy()
new_states = ["Scottland", "Jamesland", "Tamland"]

states_copy_extended.extend(new_states)

#Grabbing our last three elements which were just added
#Basically the slicing says start at index position -3, and then grab everything forward from there
print(states_copy_extended[-3:])
```

* Read documention for more functions or methods that we can use on data structures, such as append or extend, etc.

## Coding Challenge | Who will pay the bill?

```python
#Write a program that randomly chooses someone to pay the bill
#random.choice(), Return a random element from the non-empty sequence seq. If seq is empty, raises IndexError.

friends = ["Alice", "Bob", "Charlie", "David", "Emmanuel"]

random_friend = random.choice(friends)
print(f"The lucky friend who gets to pay the bill is: {random_friend}!")
```

## Nested Lists

```python
#Sometimes having all items within a same list instead of two separate lists is beneficial (fruits and veggies are somewhat related, so it'd be nice to 
#put them all in a single list)

fruits = ["Strawberries", "Nectarines", "Apples", "Grapes", "Peaches", "Cherries", "Pears"]
veggies = ["Spinach", "Kale", "Tomatoes", "Celery", "Potatoes"]

all_foods = [fruits, veggies]
print(all_foods)
```

```python
#Grabbing items from each list
print(f"First item in fruits: {all_foods[0][0]}\nFirst item in veggies: {all_foods[1][0]}")
```

## Day 4 Project | Rock Paper Scissors

```python
print("Time to play Rock, Paper, Scissors!\n")

computer_choice = random.choice(["R", "P", "S"])
human_choice = input("Do you choose rock, paper or scissors? |R, P, S| ").upper()

if human_choice not in ["R", "P", "S"]:
    print("Invalid input")
elif computer_choice == human_choice:
    print("Tie!")
elif (computer_choice == "R" and human_choice == "S") or (computer_choice == "P" and human_choice == "R") or (computer_choice == "S" and human_choice == "P"):
    print(f"Computer chose {computer_choice}. You lose!")
else: 
    print(f"Computer chose {computer_choice}. You win!")
```
