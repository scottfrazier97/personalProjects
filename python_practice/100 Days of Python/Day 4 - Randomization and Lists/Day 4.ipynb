```python
rand_num = random.randint(1, 5)

number_prompt = int(input("Pick a number between 1 and 5: "))

print(f"Computer chose: {rand_num} | You picked: {number_prompt}\n")
if rand_num == number_prompt:
    print("Great job!")
else:
    print("Incorrect! Try again")
```
