## Control Flow (If / Else) & Conditional Operators

```python
# Conditional checks with IF / ELSE

#Amusement park rider requirements
required_height = 60

rider_height_prompt = int(input("What is your height in inches? "))

if rider_height_prompt >= required_height:
    print("Great! Here is your ticket.")
else:
    print("Sorry! Not tall enough for this ride.")
```

## Modulo Operator

```python
#Returns remainders if they exist from divisions
counter = 0

print("Divisible by 2 evenly?")
while counter < 10:
    counter += 1

    if counter % 2 == 0:
        print(f"{counter} | 🆗")
    else: 
        print(f"{counter} | 🚫")
```

## Nested If Statements & Elif Statements

```python
required_height = 48

adult_age = 18
elder_age = 65

adult_price = 10
elder_price = 5
child_price = 7
mid_life_crisis = 0

rider_height_prompt = int(input("What is your height in inches? "))

if rider_height_prompt >= required_height: #Initial rider check based on height. If requirement is met, continue to next if statement below
    
    rider_age_prompt = int(input("What is your age in years? "))
    
    if rider_age_prompt >= elder_age:
        print(f"Elder tickets are ${elder_price}")
        bill = elder_price
    elif rider_age_prompt >= 45 and rider_age_prompt <= 55:
        print(f"Adult tickets are ${mid_life_crisis}")
        bill = mid_life_crisis
    elif rider_age_prompt >= adult_age:
        print(f"Adult tickets are ${adult_price}")
        bill = adult_price
    else:
        print(f"Child tickets are ${child_price}")
        bill = child_price

    photo_prompt = input("Would you like a ride photo? Y/N")

    if photo_prompt == "Y":
        print(f"Your total will be ${bill + 3}")
    else:
        print(f"Your total will be ${bill}")

else:
    print("Sorry! Not tall enough for this ride.")
```

## Coding Challenge: Pizza Orders

```python
print("Welcome!\n")

size_prompt = input("What size pizza would you like? |S, M, L| ")
pepperoni_prompt = input("Would you like pepperoni? |Y/N| ")
extra_cheese_prompt = input("Would you like extra cheese? |Y/N| ")

if size_prompt == "S":
    print(f"Small pizzas are $15")
    bill = 15
    
elif size_prompt == "M":
    print(f"Medium pizzas are $20")
    bill = 20
    
elif size_prompt == "L":
    print(f"Large pizzas are $25")
    bill = 25
    
else:
    print("Something went wrong")


if pepperoni_prompt == "Y" and size_prompt == "S":
    bill += 2
elif pepperoni_prompt =="Y" and size_prompt in ("M", "L"):
    bill += 3
else:
    pass
    
if extra_cheese_prompt == "Y":
    print(f"Your total will be ${bill + 1}")
else:
    print(f"Your total will be ${bill}")
```

## Day 3 Project | Adventure Game

```python
print("Welcome to Treasure Island! Your mission is to find the treasure!\n")

directional_prompt = input("To start things off, would you like to go left or right? |L/R| ")

if directional_prompt != "L":
    print("Whoops! You fell into a hole. Game Over.")
else:
    swim_prompt = input("Good choice! Now, would you like to swim forward or wait? |S/W| ")

    if swim_prompt != "W":
        print("Oh no! You were attacked by vicious trout hidden in the water! Game Over.")
    else:
        door_prompt = input("Great job! Now, we can see several doors in front of us (Red, Yellow, or Blue). Which one would you like to go through? |R, Y, B| ")

        if door_prompt not in ("R","Y","B"):
            print("Game Over.")
        elif door_prompt == "R":
            print("Burned by fire. Game Over.")
        elif door_prompt == "B":
            print("Eaten by beasts. Game Over")
        else:
            print("🎉You win!🎉")
```
