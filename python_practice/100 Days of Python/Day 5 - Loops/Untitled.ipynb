## For Loop

```python
#For each item in an iterable, we can perform some kind of action

fruits = ["Banana", "Peach", "Apple", "Dragonfruit", "American Elderberry"]

long_fruit_names = 0

#All indented/valid code within the for loop will be executed each time the loop runs
for item in fruits:
    if len(item) > 5:
        long_fruit_names += 1
        print(f"{item} | {len(item)} characters")
    else:
        continue #Go on to next iteration, skipping any actionable code
    
print(f"\n{long_fruit_names} out of {len(fruits)} items in the fruits list have a length longer than 5 characters")
```

## Using SUM and for loops

```python
student_scores = [100, 45, 56, 76, 53, 17, 88, 89, 95, 15, 37, 97, 43, 55, 42, 123]

total_scores = sum(student_scores)
print(f"Total of scores: {total_scores}")

max_score = 0
for score in student_scores:
    if score > max_score:
        max_score = score
    else:
        continue

print(f"Max score: {max_score}")

#Accomplishing same thing with one line of code
print(max(student_scores))
```
