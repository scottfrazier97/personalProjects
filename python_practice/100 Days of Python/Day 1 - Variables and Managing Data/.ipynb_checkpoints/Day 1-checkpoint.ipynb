## Printing

```python
#Simply prints a string (bits of text characters)
print("Hello World!")
```

## String Manipulation

```python
# Can use a docstring 
hello = """Hello World!
Hello World!
Hello World!"""
print(hello)

# or \n to print on multiple lines without multiple print statements
print("\nHello World!\nHello World!\nHello World!")
```

```python
#String concatenation (merging multiple strings into a single string)
print("Hello" + "Scott")
print("Hello" + " " + "Scott")
```

## Input Function

```python
#Allows user to input a value
name = input("What is your name? ")
age = input("What is your age? ")
favorite_hobby = input("What is your favorite hobby? ")

print(f"\nHello {name}! You are {age} years old, and your favorite hobby is {favorite_hobby}.")
```

## Python Variables

```python
#Simply assigning variable isn't enough, you need to call it
birth_year = 1997
print(birth_year)
```

```python
#Python is an interpreted language, meaning it's executed directly by an interpreter, line by line

#So we start with artist = "Illenium", and that is what the variable remains as until we reassign that same variable name later to "Zedd"
artist = "Illenium"
length = len(artist)
print(f"{artist} | {length} characters")

artist = "Zedd"
length = len(artist)
print(f"{artist} | {length} characters")
```
