## Functions

```python
#Defining and calling functions. 
#Functions have a name followed by parenthesis...print()
#Allows us to cut down on repeatable code

#Creating our own functions requires the def keyword, and code inside must be indented
def addition_func_sq(a, b):
    return (a + b) ** 2

#Functions require us to call them to actually run
addition_func_sq(10, 52)
```

## While Loops

```python
#Continues running while a condition is true
sleeping_hours = 8

while sleeping_hours > 0:
    sleeping_hours -= 1
    print("Getting a much needed rest.")

if sleeping_hours == 0:
    print("Time to wake up!")
else:
    pass
```
