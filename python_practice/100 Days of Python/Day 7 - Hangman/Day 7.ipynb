## Breaking down complex problems into a flow chart

## Step 1

### Picking a random word

```python
#Computer chooses a word
import random

word_choices = ['kayak', 'memento', 'swivel', 'yacht', 'skyscraper', 'mystify', 'luxury', 'familiar', 'aloof', 'patience']
computer_choice = random.choice(word_choices)
print(computer_choice)
```

### Asking user for a letter

```python
user_letter = input("Please enter your guess: ").lower()
```

### Check if letter is present in word

```python
for letter in computer_choice:
    if letter in user_letter:
        print("Present")
    else:
        print("Absent")
```
