## Combinatorics Intro

* Often referred to as the study of counting
* Helps us count things when the results might not be immediately clear
* Many real life business challenges and data science decisions are directly related to the ability to count possible combinations or permutations of objects
* Possible IRL application: `"For a subscription box company, how many possible combinations of products are there that can be delivered to a customer?"`

## Factorial

* In math, the factorial is an operation on a non-negative number integer <b>n</b> that is equal to the product of all positive integers less than or equal to <b>n</b>.
* Product of all positive integers less than or equal to n
