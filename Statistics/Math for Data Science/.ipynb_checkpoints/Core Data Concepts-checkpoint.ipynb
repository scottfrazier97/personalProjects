# Intro to Core Data Concepts

What is data?
* Collected observations and information about something which can be structured or unstructured
* Continuous: A set of data is said to be continuous if the values belonging to the set can take on any value within a finite or infinite interval.
* Discrete: A set of data is said to be discrete if the values belonging to the set are distinct and separate (no in-between values)
* Structured
