# Intro to Core Data Concepts

What is data?
* Collected observations and information about something which can be structured or unstructured
* <b>Continuous (Numeric only)</b>: A set of data is said to be continuous if the values belonging to the set can take on any value within a finite or infinite interval.
* <b>Discrete (Numeric or String)</b>: A set of data is said to be discrete if the values belonging to the set are distinct and separate (no in-between values)
* <b>Nominal</b>: Classified without a natural order or rank (Categories of animals). A good test for nominal vs not nominal is to see if the data can be clearly sorted or not. Nominal data <b>cannot be sorted.</b>
* <b>Ordinal</b>: Data that can be sorted (it has an <b>order</b> to it, i.e.: Airline passenger classes is an ordinal discrete data set). Doesn't necessarily need to be numeric...Hot, mild, cold is categorical but can also be sorted logically.
