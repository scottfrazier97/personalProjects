# Intro to Core Data Concepts

## What is data?
* Collected observations and information about something which can be structured or unstructured
* <b>Continuous (Numeric only)</b>: A set of data is said to be continuous if the values belonging to the set can take on any value within a finite or infinite interval.
* <b>Discrete (Numeric or String)</b>: A set of data is said to be discrete if the values belonging to the set are distinct and separate (no in-between values)
* <b>Nominal</b>: Classified without a natural order or rank (Categories of animals). A good test for nominal vs not nominal is to see if the data can be clearly sorted or not. Nominal data <b>cannot be sorted.</b>
* <b>Ordinal</b>: Data that can be sorted (it has an <b>order</b> to it, i.e.: Airline passenger classes is an ordinal discrete data set). Doesn't necessarily need to be numeric...Hot, mild, cold is categorical but can also be sorted logically.
* <b>Structured</b>: Highly specific and is stored in a predefined format. (Excel spreadsheets, JSON, XML, SQL Databases)
* <b>Unstructured</b>: Not in a particular format. (Audio, video, text data doesn't need to follow any particular predefined structured format)
  * Typically harder to work with but sometimes is necessary to achieve the desired result.
* <b>Population vs Sample</b>: Need to consider the scope of our data collection. Is this data a full representation of everything available and known? Or is it a sample of everything?
  * <b>Population (N)</b>: Consists of every member of a group. This is dependent on the context of the situation. (A list of all student names in a school contains data on the population of the school).
  * <b>Sample (n)</b>: Smaller, specific subset of the population from which you actually collect data. Used to make inferences (educated guesses) about the larger population when studying the whole group is not practical. Best to be sure that the sample is representative of your population.

# Measurements of Central Tendency

## Mean
* (sum of data points / number of data points)
* Population Mean (mu: μ): Mean of entire population
* Sample Mean (x-bar: x̄): Mean of sample

Weighted Mean
* The weighted arithmetic mean is similar to an ordinary arithmetic mean, except that instead of each of the data points contributing equally to the final average, some data points contribute more than others.

Truncated Mean
* It involves the calculation of the mean after discarding given parts of a probability distribution or sample at the high and low end, and typically discarding an equal amount of both.
* Could be getting rid of an outlier if that value is too large, but we would also have to get rid of a value at the low end of the data set too.

```python
def mean_finder(data):

    final_mean = sum(data) / len(data)
    return final_mean

result = mean_finder([1,2,3,4,5])
print(f"Mean: {result}")

#-----------------WEIGHTED MEAN-----------------
def weighted_mean(dept_size, means):

    weights_products = sum([x * y for x, y in zip(dept_size, means)])
    total_dept_count = sum(dept_size)

    final_weighted_mean = weights_products / total_dept_count
    return final_weighted_mean

result2 = round(weighted_mean([20,7,13,25], [8.4,6.1,9.1,7.8]), 1)
print(f"Weighted Mean: {result2}")

#-----------------TRUNCATED MEAN-----------------
def truncated_mean(data2):

    truncated_list = [n for n in data2 if n != min(data2) and n!= max(data2)]
    final_truncated_mean = sum(truncated_list) / len(truncated_list)

    return final_truncated_mean
    
result3 = round(truncated_mean([16,18,21,27,32,32,33,91]), 1)
print(f"Truncated Mean: {result3}")  
```

## Mode

* The value that appears the <b>most often</b> in our dataset.
* Bi-modal data is when we have two values that appear the most, but also an equal number of times
* No mode is when all values appear the same time

```python
import statistics

data = [1, 2, 2, 3, 4, 4, 4, 5]
mode_value = statistics.mode(data)
print(f"Mode: {mode_value}")
```

## Median

* The <b>middle point</b> of our ordered dataset
* If we have two middle points (even number of data points), we sum them together, and divide by 2. The result is our median.

```python
data = [329,479,459,269,369,799,649]
median_value = statistics.median(data)
print(f"Median: {median_value}")
```

# Measurements of Dispersion
* Set out to explain how far our data is spread out <b>around the mean</b>.
* The smaller our end result for variance and standard deviation, the more close together our values are around the mean.
* We use the mean for variance because the mean is the value that minimizes the sum of squared deviations, making it the most natural measure of central tendency to use for calculating spread

## Variance
* <b>Population Variance (σ²)</b>
  * Formula: `σ² = Σ(xᵢ - μ)² / N`
  * Mean of the squared deviations
* <b>Sample Variance (s²)</b>
  * Formula: `s² = Σ(xᵢ - x̄)² / (n - 1)`
  * Same as population variance but we use X-bar in our formula instead of mu, and we use n-1 instead of just N in the denominator.
    * Using the sample mean to compute variance makes the raw variance too small.
    * Dividing by n–1 corrects this downward bias.
    * You lose one degree of freedom because the mean is estimated via it being a sample to begin with.
    * Therefore sample variance uses n–1 to give an unbiased estimate of the true variance.  
* We square the result of (data point - mean) because we might end up with negative numbers from our subtraction. <b>We want all positive.</b>
* We also get to magnify the larger deviations which has a larger effect on the variance than smaller deviations, which helps illustrate a better picture of how dispersed our values are.'
* Hard to interpret because our units become so large and are no longer in the original unit of measurement for what we are analyzing. 

## Standard Deviation
* <b>Population Variance (σ)</b> is square root of population variance.
* <b>Sample Variance (s)</b> is square root of sample variance.
* We want to take the square root because it <b>puts our units back into an understandable format.</b>
  * Ex: So if the average house price is $$350,000 and the std dev is $50,000, people can immediately understand:
  * “Most house prices tend to be within about $50,000 of the average.”
  * Which is much more intuitive than saying the variance is 2.5 × 10,000,000,000 dollars squared.

```python
import numpy as np

def summary_stats(numbers):
    squares = []
    
    nums_mean = sum(numbers) / len(numbers)
    
    for num in numbers:
        squared_result = (num - nums_mean) ** 2
        squares.append(squared_result)

    pop_variance = sum(squares) / len(numbers)
    sampl_variance = sum(squares) / (len(numbers) - 1) #Same as population variance, but in our denominator we divide by is (n - 1)

    pop_std_dev = np.sqrt(pop_variance)
    sampl_std_dev = np.sqrt(sampl_variance)

    print(f"Mean = {round(nums_mean, 2)}")
    print(f"Population variance = {round(pop_variance, 2)}")
    print(f"Sample variance = {round(sampl_variance, 2)}")
    print(f"Population Std Dev = {round(pop_std_dev, 2)}")
    print(f"Sample Std Dev = {round(sampl_std_dev, 2)}")

summary_stats([220.1,220.4,220.2,220.0,220.1])
```

# Quartiles and IQR

## Quartiles
* Centered around median
* Split our data into three quartiles
  * First quartile (Q1, Lower quartile, 25th percentile): Median of our lower half of data (before actual median)
  * Second quartile (Q2, Second median, 50th percentile): True median of our data
  * Third quartile (Q3, Upper quartile, 75th percentile): Median of our upper half of data (after actual median)
* Our quartiles give us our Five Number Summary
  * Minimum > Q1 > Median > Q2 > Maximum
  * Range: Max - Min
  * Everything between Q1 and Q3 represent that middle 50th percentile
  * If we have an odd number of values, we can exclude the original median so we may calculate Q1 and Q3

## Inner Quartile Range
* Formula: `IQR = Q3 - Q1`
* This gives us an idea of how spread out our values are around the median
* Identifying Outliers:
  * Any value in the set of data that is lower than: `Q1 - 1.5(IQR)`
  * Any value in the set of data that is larger than: `Q3 + 1.5(IQR)`

```python
def find_iqr_outliers(data):

    import pandas as pd

    df = pd.DataFrame(data)
    df_new = df.rename(columns={0: 'Height'})
    
    df_describe = df_new.describe()
    
    q1_outliers_threshold = df_describe.loc['25%'].item()
    q3_outliers_threshold = df_describe.loc['75%'].item()
    
    inner_quartile_range = q3_outliers_threshold - q1_outliers_threshold
    
    q1_outliers = [n for n in data if n < (q1_outliers_threshold - (1.5 * inner_quartile_range))]
    q3_outliers = [n for n in data if n > (q3_outliers_threshold + (1.5 * inner_quartile_range))]

    return q1_outliers_threshold, q3_outliers_threshold, q1_outliers, q3_outliers

q1_threshold, q3_threshold, q1_result, q3_result = find_iqr_outliers([66,67,67,68,68,68,69,69,69,69,70,70,71,71,72,73,75])

if len(q1_result) == 0 and len(q3_result) == 0:
    print("No outliers found")
else:
    print(f"Any number less than {q1_threshold} or greater than {q3_threshold} is an outlier.\n\nLower outliers: {q1_result}\nUpper outliers: {q3_result}")
```
