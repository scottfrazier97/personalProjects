# Visualizing Data

## Intro to Visualization

* Always ask, "What is the information I am wanting to share or story I am trying to tell? How does this visualization help in conveying that information to others?"
* Crucial in decision making based on our final data or result shown in the visualization.
* Key part: Understanding what type of plot, chart, or visualization to use.

## Scatter Plot

* A typical scatter plot will represent <b>two dimensions</b> (data features).
* Try to show as <b>few features as possible</b> to get your point across.
* Think about putting the independent variable along the x-axis, and the dependent variable along the y-axis.
* Keep in mind the correlation does not equal causation.
* Both variables should also be continuous.

```python
import seaborn as sns

data = sns.load_dataset(name='penguins')
data
```

```python
sns.scatterplot(data, x='body_mass_g', y='bill_depth_mm', hue='species')
```

* Regression line can be simplified into something like that of the equation of a slope
  * `ŷ = mx + b`
    * y-hat = We are specifically discussing a trend line.
    * m = slope of the line. Tells us the rate at which the line is increasing or decreasing.
      * A higher value of m indicates a line that is steeper.
      * A lower value of m indicates a line that is more horizontal.
      * A negative value of m, indicates that the line is trending downwards.
    * b = the point at which our line intersects the y-axis.
    * If we have the values for `m and b`, then we can plug in any value for `x` and it will give us the value for `y`
* Regression line is meant to be a line of best fit.
  * Also meant to be used as a <b>guideline</b> to determine a value for our `y` when we pick a particular value for `x`
  * For example, if we use `x = 15` on our graph below, we can guesstimate that for a bill_depth_mm value of 15, we will see a value of around `flipper_length_mm = 217` when look at the corresponding spot on the y-axis.
  * This is NOT meant to be used as true predictions. Just an estimate of what we can expect on average.

```python
# Seaborn regression plot with regression line showing the trend of our data. 
gentoo_pengs = data[data['species'] == 'Gentoo']

sns.regplot(gentoo_pengs, x='bill_depth_mm', y='flipper_length_mm')
```

## Line Plots

* Only use line plots when we know a correlation (continuous relationship) to exist between the two data features.
* Especially useful for variables that are time oriented.
* X-axis values should be something that makes sense to track over time. E.g.: Number of passengers over the years.
* Can also use an ogive plot which shows the cumulative sum of our variable up to that point in time.
  * Only makes sense for certain type of data like revenue, or a count.
  * Would not make sense for something like tracking temperature over time.

```python
flights = sns.load_dataset(name='flights')
flights
```

```python
#Takeaway: Summer months are always busy flight months year over year.
sns.lineplot(flights, x='year', y='passengers', hue='month')
```

```python
flights_1960 = flights[flights['year'] == 1960]

sns.lineplot(flights_1960, x='month', y='passengers')
```

## Distribution Plots

* Usefule for measuring the <b>dispersion</b> of our data. Refer to measurements of dispersion (variance and std dev) from first notebook.
* One of the most common plots is a <b>histogram</b>.
  * X-axis is <b>continuous</b>
  * Histograms have no spaces between bars due to the continuous nature of the histogram variables.
  * Each bar shows count of that bin.
  * Bin widths must be the same so that the distribution is consistent across the data.
  * Choosing the number of buckets is more of an art than an exact science.
    * Choose a bin number the properly fits the data we are working with.
  * If we were to add a point in the center-top of each bar, and then connect them with a line, we would get a probability distribution.

```python
mpg = sns.load_dataset(name='mpg')
mpg
```

```python
sns.histplot(mpg, x='mpg', bins=30, hue='origin')
```

## Categorical Plots

* Display some <b>metric per category</b>
* Bar plots are common for this purpose
* Have space between bars to indicate this is categorical data we are dealing with on the x-axis.
* Great for comparing two different variables.

```python
sns.barplot(flights, x='month', y='passengers', hue='month')
```

```python
sns.countplot(mpg, x='origin', hue='origin')
```

```python
sns.barplot(data, x='sex', y='bill_depth_mm', hue='species', palette='mako')
```

### Boxplot

* Great way to visualize the full distribution of our data using the five number summary
* Line inside box is the median, Q1 is the left end of the box and Q2 is the right end of the box, and our min is the far left line with our max being the far right line

```python
sns.boxplot(mpg, x='horsepower', y='origin', hue='origin')
```

### KDE Plot

* Regarding the kernel density function, we just want to think about it's width and height
  * Width = bandwidth, Height = amplitude
  * Bandwidth is how we dictate how smooth the KDE line will appear
    * A lower value will indicate a less-smooth curve
  * Essentially a way for us to see the distribution with a smooth line as opposed to the bars present in a histogram
    * Can also combine it with a histogram though.

```python
#
sns.histplot(data, x='body_mass_g', kde=True)
```

```python
sns.kdeplot(data, x='body_mass_g', bw_adjust=0.5)
```

### Violin Plot

* Taking a KDE plot and creating a mirror image of it along it's axis.
* Line along the box plot do not represent the whiskers from box plot. They represent a `95% confidence interval`

```python
sns.violinplot(data, x='body_mass_g', y='species', hue='species')
```

## Common Plot Pitfalls

* Misleading Axes
  * Want your scales to be correct, and proportionate to the visualization being used.
  * Usually best to start axes at 0 to curb this.
  * Need to keep the scale consistent throughout the axis
  * Keep spacing consistent with labels.
* Proper titles and labels
  * Data labels on top of bars in a bar graph can help with clarity.
