# Recommender Systems

### Theory

* Two most common types of recommender systems are Content-Based and Collaborative Filtering
  * Collaberative Filtering produces recommendations based on the knowledge of the users attitude towards items, that is it uses the "wisdom of the crowd" to recommend items
  * Content-Based recommender systems focus on the attributes of the items and give you recommendations based on the similarity between them
  * CF is more commonly used because it usually gives better results and is easier to understand
  * Algorithm has ability to do feature learning on it's own, which means that it can start to learn for itself what features to use
  * CF can also be divided into 2 types of filtering
    * Memory-Based Collaberative Filtering
    * Model-Based Collaberative Filtering

### Model Building (Item similarity based recommender)

```python
import numpy as np
import pandas as pd
```

```python
columns_names = ['user_id', 'item_id', 'rating', 'timestamp']
```

```python
df = pd.read_csv("u.data", sep='\t', names=columns_names)
```

```python
df.head()
```

```python
movie_titles = pd.read_csv('Movie_Id_Titles')
```

```python
movie_titles.head()
```

```python
df = pd.merge(df, movie_titles, on="item_id")
```

```python
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.groupby('title')['rating'].mean().sort_values(ascending=False).head()
```

```python
df.groupby('title')['rating'].count().sort_values(ascending=False).head()
```

```python
ratings = pd.DataFrame(df.groupby('title')['rating'].mean())
```

```python
ratings.head()
```

```python
ratings['num of ratings'] = df.groupby('title')['rating'].count()
ratings.head()
```

```python
sns.histplot(ratings, x='num of ratings', bins=45)
```

```python
sns.histplot(ratings, x='rating', bins=45)
```

```python
#Appears that the more ratings a movie has, the higher the rating typically
sns.jointplot(x='rating', y='num of ratings', data=ratings, alpha=0.5)
```

```python
moviemat = df.pivot_table(index='user_id', columns='title', values='rating')
moviemat
```

```python
ratings.sort_values('num of ratings', ascending=False).head(10)
```

```python
starwars_user_ratings = moviemat['Star Wars (1977)']
liar_liar_ratings = moviemat['Liar Liar (1997)']
```

##### Star Wars

```python
#Shows if a user rated a movie or not
starwars_user_ratings.head()
```

```python
#Corrwith uses two dataframe objects
similar_to_starwars = moviemat.corrwith(starwars_user_ratings)
```

```python
similar_to_liar = moviemat.corrwith(liar_liar_ratings)
```

```python
#Shows us how correlated the user ratings of other movies are to the user ratings of Star Wars
corr_star_wars = pd.DataFrame(similar_to_starwars, columns = ['Correlation'])
corr_star_wars.dropna(inplace=True)
corr_star_wars.head()
```

```python
#Perfect correlation, but this does not make sense. Need to filter out movies with less than a certain number of reviews.
corr_star_wars.sort_values('Correlation', ascending=False).head(10)
```

```python
#Join to get num of ratings, and then filter out unnecessary movies
corr_star_wars = corr_star_wars.join(ratings['num of ratings'])
```

```python
# These results now make much more sense
# Austin powers sees a fairly large drop in correlation, but it might be showing up here because it is a popular movie, and so is Star Wars
# Therefore, if you like popular movies, then the system will recommend popular movies
corr_star_wars[corr_star_wars['num of ratings'] > 100].sort_values('Correlation', ascending=False).head()
```
