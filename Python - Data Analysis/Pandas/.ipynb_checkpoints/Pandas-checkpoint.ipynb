```python
import pandas as pd
import numpy as np
```

### Series

```python
labels = ['a','b','c']
my_data = [10,20,30]
arr = np.array(my_data)
d = {'a': 10,
     'b': 20,
     'c': 30}
```

```python
pd.Series(data = my_data)
```

```python
pd.Series(data = my_data, index=labels) #Using specified parameters
```

```python
pd.Series(my_data, labels) #Using ordinal position of parameters, no need to specify parameters if you place in proper order
```

```python
pd.Series(arr) #NumPy array as argument to be passed in
```

```python
pd.Series(arr, labels) #Assigning labels with NumPy array
```

```python
pd.Series(d) #Using a dictionary, which assigns the keys as the index and values as the values
```

```python
pd.Series(arr)
```

```python
ser1 = pd.Series([1,2,3,4], ['USA', 'Germany', 'USSR', 'Japan'])
ser1
```

```python
ser1['USA'] #Grabbing data point for the USA index, typing it as a string because we know USA is a string
```

```python
ser2 = pd.Series([1,2,5,4], ['USA', 'Germany', 'Italy', 'Japan'])
ser2
```

```python
ser3 = pd.Series(data = labels) #Grabbing information with integer index
ser3[0]
```

```python
ser1 + ser2 #Adding the two series together, it tries to match it on the index, returns NaN
```
