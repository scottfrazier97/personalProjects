### Series

```python
labels = ['a','b','c']
my_data = [10,20,30]
arr = np.array(my_data)
d = {'a': 10,
     'b': 20,
     'c': 30}
```

```python
pd.Series(data = my_data)
```

```python
pd.Series(data = my_data, index=labels) #Using specified parameters
```

```python
pd.Series(my_data, labels) #Using ordinal position of parameters, no need to specify parameters if you place in proper order
```

```python
pd.Series(arr) #NumPy array as argument to be passed in
```

```python
pd.Series(arr, labels) #Assigning labels with NumPy array
```

```python
pd.Series(d) #Using a dictionary, which assigns the keys as the index and values as the values
```

```python
pd.Series(arr)
```

```python
ser1 = pd.Series([1,2,3,4], ['USA', 'Germany', 'USSR', 'Japan'])
ser1
```

```python
ser1['USA'] #Grabbing data point for the USA index, typing it as a string because we know USA is a string
```

```python
ser2 = pd.Series([1,2,5,4], ['USA', 'Germany', 'Italy', 'Japan'])
ser2
```

```python
ser3 = pd.Series(data = labels) #Grabbing information with integer index
ser3[0]
```

```python
ser1 + ser2 #Adding the two series together, it tries to match it on the index, returns NaN
```

### DataFrames

```python
from numpy.random import randn
```

```python
np.random.seed(101)
```

```python
df = pd.DataFrame(randn(5,4),['A', 'B', 'C','D','E'],['W','X','Y','Z'])
df
```

##### Selecting Columns

```python
df['W'] #OR df.W
```

```python
df[['W','Z',]] #Selecting multiple columns
```

```python
df['New'] = df['W'] + df['Y'] #Creating a new column
df
```

##### Selecting Rows

```python
df.loc['A'] #loc stands for location, and takes in a index label
```

```python
df.iloc[0] #iloc stands for index based location, and takes in a numerical index location
```

##### Selecting Specific Values

```python
df.loc['B','Y'] #Returning a value located at row, and column location (Row B, Column Y)
```

```python
df.loc[['A', 'B'], ['W', 'Y']]
```

##### Dropping Columns

```python
#Deleting a column, need to specify the axis = 1 for COLUMNS, defaults to axis = 0 which is for INDEX
#Syntax below does not edit DF inplace (meaning it only drops in this view, and does not edit the original DF)
#Use the inplace = True argument to accomplish this

df.drop('New', axis=1, inplace=True)
df
```

##### Dropping Rows

```python
df.drop('E', axis=0) #Axis=0 is the default, so not necessary to specify in this case
```

```python
df.shape #Checking the number of columns and rows (Also why rows are referred to as the 0 axis...(rows, columns))
```

##### Conditional Selection

```python
df > 0
```

```python
df[df>0]
```

```python
df['W'] > 0
```

```python
df[df['W'] > 0] #Since we are passing in a series, we no longer get the null values
```

```python
df[df['Z'] < 0]
```

```python
df[df['W']>0]['X'] #Return values for X column where W column is > 0
```

```python
#Using and instead of & will result in an ambigous error, meaning that pandas/python gets confused
#when you try to pass more than one boolean value into the and comparison...use the pipe operator for OR...

df[(df['W']>0) & (df['Y']>1)] 
```

```python
df.reset_index() #Resetting the index to a numerical value, getting rid of the alpha characters, inplace=True to fully overwrite
```

```python
newind = 'CA NY WY OR CO'.split()
newind
```

```python
df['States'] = newind
df
```

```python
df.set_index('States') #Assigning new index values based on the States columns we just created
```

##### MultiIndex

```python
#Index levels
outside = ['G1', 'G1','G1','G2','G2','G2']
inside = [1,2,3,1,2,3]
hier_index = list(zip(outside,inside))
hier_index = pd.MultiIndex.from_tuples(hier_index)
hier_index
```

```python
df = pd.DataFrame(randn(6,2),hier_index,['A','B'])
df
```

```python
df.loc['G1'] #call from outside index, and then contine to go deeper
```

```python
df.loc['G1'].loc[1] #call from outside index, and then contine to go deeper
```

```python
df.index.names = ['Groups','Num']
df
```

```python
df.loc['G1'].loc[2]['A']
```

```python
df.loc['G1']
```

```python
#Cross Section for Multi Index...Grabbing values for Num index = 1 in G1 and G2
df.xs(1, level = 'Num')
```
