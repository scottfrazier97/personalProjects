# Logistic Regression

Now it's time to do a train test split, and train our model!

You'll have the freedom here to choose columns that you want to train on!

** Split the data into training set and testing set using train_test_split**

```python
ad_data
```

```python
ad_data.columns
```

```python
#Given that the categorical variables have so many distinct values, this would likely lead to too many additional columns via encoding
#Going to leave them out of the analysis for now
for col in ['Ad Topic Line', 'City', 'Country']:
    print(f"{col} has {ad_data[col].nunique()} distinct values")
```

```python
ad_data.drop(['Ad Topic Line', 'City', 'Country','Timestamp'], axis=1, inplace=True)
ad_data
```

```python
#Assigning predictor and target variables
X = ad_data.drop('Clicked on Ad', axis=1)
y = ad_data['Clicked on Ad']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

** Train and fit a logistic regression model on the training set.**

```python
from sklearn.linear_model import LogisticRegression
logmodel = LogisticRegression(max_iter=2000)
```

```python
logmodel.fit(X_train, y_train)
```

## Predictions and Evaluations
** Now predict values for the testing data.**

```python
predictions = logmodel.predict(X_test)
```

** Create a classification report for the model.**

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test, predictions))
```

```python
print(confusion_matrix(y_test, predictions))
```

<b>Class 0:</b> <br>

* <b>Precision (0.96):</b> When the model predicts class 0, it's correct 96% of the time.

* <b>Recall (0.97):</b> The model correctly identifies 97% of all actual class 0 instances.

* <b>F1-score (0.97):</b> Harmonic mean of precision and recall. Indicates a strong balance.

* <b>Support (146):</b> There are 146 actual examples of class 0 in the dataset.

<b>Class 1:</b> 

* <b>Precision (0.97):</b> When the model predicts class 1, it's correct 97% of the time. 

* <b>Recall (0.96):</b> The model only captures 96% of the actual class 1 instances—some are missed. 

* <b>F1-score (0.97):</b> Lower than class 0, due to the lower recall.

* <b>Support (154):</b> Fewer examples of class 1 (possibly causing slight imbalance). 

<b>Overall</b> <br>
* <b>Accuracy (0.97):</b> The model correctly predicts the class in 97% of all cases—excellent overall performance.
* <b>Macro avg (0.97):</b> Precision, recall, and F1 are evenly averaged across both classes, confirming consistent performance.
* <b>Weighted avg (0.97):</b> Performance remains strong even when accounting for the slightly different class sizes
* The model performs equally well on both class 0 and class 1, with F1-scores of 0.97 for each
* There’s no significant imbalance in support, and no clear weakness in either class’s recall or precision

```python
logmodel.coef_
```

```python
cdf = pd.DataFrame(logmodel.coef_.T, X.columns, columns=['Coeff']).sort_values(by='Coeff', ascending=False)
cdf
```

### Interpreting Coefficients

<b>Age (0.177573)</b>
* A 1-year increase in <b>Age</b> increases the log-odds of clicking on an ad by 0.177573.
* This means that as people age, we have a higher likelihood of them clicking on the ad.
* Given our outcome value, Age seems to be a strong predictor here.
  
<b>Area Income (-0.000148):</b>
* A $1 increase in <b>Avg Area Income</b> slightly decreases the log-odds of clicking on an ad by -0.000148.
* The association of an individual being from a wealthy area has an ever so slightly negative affect on the log-odds of someone clicking on an ad.  

<b>Daily Internet Usage (-0.071801):</b> 
* A 1-minute increase in <b>Avg Daily Internet Usage</b> decreases the log-odds of clicking on an ad by -0.071801.
* This means those who spend more time on the internet are less likely to click on an ad.

<b>Daily Time Spent on Site (-0.191964):</b> 
* A 1-minute increase in <b>Daily Time Spent on Site</b> decreases the log-odds of clicking on an ad by -0.191964.
* As the amount of time an individual spends on the website increases, we see a decrease in the likelihood of them clicking on an ad.

<b>Male (-0.759926):</b> 
* Being <b>Male</b> decreases the log-odds of clicking on an ad by -0.759926.
* All other variables being equal, Males are much less likely to click on an ad than females.
