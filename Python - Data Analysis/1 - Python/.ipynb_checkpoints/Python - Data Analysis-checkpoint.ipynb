## map and filter

```python
seq = [1,2,3,4,5]
```

Can use map to apply a function to every item in an iterable. Below example applies times2 function to each item in seq list, returning each item times 2

```python
map(times2,seq)
```

```python
list(map(times2,seq))
```

Below is the lambda re-written version of our times2 function. Attempting to make function all on one line, getting rid of keywords we don't need

```python
list(map(lambda var: var*2,seq))
```

Filter has a similar structure to map, but you pass in a function or lambda expression and filters out boolean values

```python
filter(lambda item: item%2 == 0,seq)
```

```python
list(filter(lambda item: item%2 == 0,seq))
```
