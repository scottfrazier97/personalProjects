```python
df3.head()
```

```python
df3.plot.scatter(x='a', y='b', figsize=(12,3), c='red', s=30)
```

** Create a histogram of the 'a' column.**

```python
df3['a'].hist()
```

```python
import seaborn as sns
sns.set_style()
df3['a'].hist(bins=25)
```

** Create a boxplot comparing the a and b columns.**

```python
df3[['a','b']].plot.box()
```

** Create a kde plot of the 'd' column **

```python
df3['d'].plot.kde(lw=3, ls='--', c='red')
```

** Create an area plot of all the columns for just the rows up to 30

```python
df3.head(30).plot.area()
```
