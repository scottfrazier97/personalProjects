### Series

##### MultiIndex

```python
#Index levels
outside = ['G1', 'G1','G1','G2','G2','G2']
inside = [1,2,3,1,2,3]
hier_index = list(zip(outside,inside))
hier_index = pd.MultiIndex.from_tuples(hier_index)
hier_index
```

```python
df = pd.DataFrame(randn(6,2),hier_index,['A','B'])
df
```

```python
df.loc['G1'] #call from outside index, and then contine to go deeper
```

```python
df.loc['G1'].loc[1] #call from outside index, and then contine to go deeper
```

```python
df.index.names = ['Groups','Num']
df
```

```python
df.loc['G1'].loc[2]['A']
```

```python
df.loc['G1']
```

```python
#Cross Section for Multi Index...Grabbing values for Num index = 1 in G1 and G2
df.xs(1, level = 'Num')
```

### Missing Data

```python
d = {'A':[1,2,np.nan],'B':[5,np.nan,np.nan],'C':[1,2,3]}
df = pd.DataFrame(d)
df
```

##### Dropping

```python
#If any rows have a null value, it drops it from the dataset. 

df.dropna() #Defaults to axis=0 which is rows
```

```python
df.dropna(axis = 1) #Defaults to axis=0 which is rows, use axis=1 for columns
```

```python
#thresh parameter can be used to keep specific rows/columns based on how many nulls there are. Threshold specified must be met
#to drop those rows/columns
df.dropna(thresh=2)
```

##### Replacing

```python
df.fillna(value='Fill VALUE')
```

```python
df['A'].fillna(value=df['A'].mean())
```

### Groupby

```python
data = {'Company': ['Goog','Goog','MSFT','MSFT','FB','FB'],
        'Person': ['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
        'Sales': [200,120,340,124,243,350]}
companies_df = pd.DataFrame(data)
companies_df
```

```python
companies_df.groupby('Company') #Returns a groupby object
```

```python
byCompany = companies_df.groupby('Company') #Assign the groupby object to a variable (still groupby object)
byCompany.mean() #Perform an aggregate function on the group
```

```python
byCompany.sum()
```

```python
companies_df.groupby('Company').sum().loc['FB']
```

```python
companies_df.groupby('Company').count()
```

```python
companies_df.groupby('Company').max()
```

```python
companies_df.groupby('Company').describe()
```

### Merging, Joining, Concatenating

```python
df1 = pd.DataFrame({'A': ['A0', 'A1', 'A2', 'A3'],
                        'B': ['B0', 'B1', 'B2', 'B3'],
                        'C': ['C0', 'C1', 'C2', 'C3'],
                        'D': ['D0', 'D1', 'D2', 'D3']},
                        index=[0, 1, 2, 3])
```

```python
df2 = pd.DataFrame({'A': ['A4', 'A5', 'A6', 'A7'],
                        'B': ['B4', 'B5', 'B6', 'B7'],
                        'C': ['C4', 'C5', 'C6', 'C7'],
                        'D': ['D4', 'D5', 'D6', 'D7']},
                         index=[4, 5, 6, 7]) 
```

```python
df3 = pd.DataFrame({'A': ['A8', 'A9', 'A10', 'A11'],
                        'B': ['B8', 'B9', 'B10', 'B11'],
                        'C': ['C8', 'C9', 'C10', 'C11'],
                        'D': ['D8', 'D9', 'D10', 'D11']},
                        index=[8, 9, 10, 11])
```

##### Pandas Concatenation

```python
#Concatenation glues together dataframes. Dimensions of the DFs should match along the axis you are concatenating on.
#Default axis to join on is the rows
pd.concat([df1,df2,df3])
```

##### Merging

```python
left = pd.DataFrame({'AccountNumber': ['001', '002', '003', '004'],
                     'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3']})
   
right = pd.DataFrame({'AccountNumber': ['001', '002', '003', '004'],
                          'C': ['C0', 'C1', 'C2', 'C3'],
                          'D': ['D0', 'D1', 'D2', 'D3']})    
```

```python
pd.merge(left,right,how='inner',on='AccountNumber')
```

##### Joining

```python
#Combining the columns of two potentially differently indexed dataframes, into a single resulting DataFrame
left2 = pd.DataFrame({'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3']},
                      index = ['001', '002', '003', '004'],)
   
right2 = pd.DataFrame({'C': ['C0', 'C1', 'C2', 'C3'],
                      'D': ['D0', 'D1', 'D2', 'D3']},
                      index =  ['001', '002', '003', '004'])
```

```python
left2.join(right2)
```
