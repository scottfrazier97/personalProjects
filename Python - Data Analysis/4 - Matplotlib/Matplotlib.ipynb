# Matplotlib

[Matplotlib Documentation](https://matplotlib.org/stable/index.html)

```python
import matplotlib.pyplot as plt
```

```python
#Seeing the plots we create in jupyter notebook, use plt.show() if not in jupyter notebook
%matplotlib inline
```

```python
import numpy as np
x = np.linspace(0,5,11)
y = x ** 2
```

```python
x
```

```python
y
```

### Functional Method

```python
plt.plot(x,y) # Add 'r-' as an argument after y for the color and line style
plt.xlabel('This is my X label')
plt.ylabel('This is my Y label')
plt.title('This is my Title')
```

##### Multi-plots on the same canvas

```python
plt.subplot(1,2,1) #Number of rows, columns, and plot number we are referring to
plt.plot(x,y,'r') #Color red with 'r'

plt.subplot(1,2,2)
plt.plot(y,x, 'b')
```

### Object Oriented API Method

```python
#The main idea is to create figure objects, and then call methods off of it

#Think of this as an imaginary blank canvas to begin plots with
fig = plt.figure() 

#Add in axes by passing in a list, takes four arguments: left, bottom, width, height...range from 0 to 1
axes = fig.add_axes([0.1,0.1,0.8,0.8]) 

axes.plot(x,y)
axes.set_xlabel('OO X Label')
axes.set_ylabel('OO Y Label')
axes.set_title('OO Title')
```

##### Manually Setting Axes

```python
fig2 = plt.figure()

axes1 = fig2.add_axes([0.1,0.1,0.8,0.8])
axes2 = fig2.add_axes([0.2,0.5,0.4,0.3])

axes1.plot(x,y)
axes1.set_title('LARGER')

axes2.plot(y,x)
axes2.set_title('smaller')
```

##### Automatically Setting Axes

```python
#Subplots lets you specify the number of rows and columns
#Using tuple unpacking below, allows us to instantiate initial figure object, as well as automatically take care of axes too
fig, axes = plt.subplots(nrows = 1, ncols = 2) 

#axes is simply an array of matplotlib axes objects, which means we can iterate through it
for current_ax in axes:
    current_ax.plot(x,y)
```

```python
#We can also index in addition to iterate
fig, axes = plt.subplots(nrows = 1, ncols = 2) 

axes[0].plot(x,y)
axes[0].set_title('First Plot')

axes[1].plot(y,x)
axes[1].set_title('Second Plot')

#Fixing Overlapping Plots
plt.tight_layout()
```

### Figure Size & DPI

[Legend Documentation](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.legend.html#matplotlib.axes.Axes.legend)

```python
fig = plt.figure(figsize=(4,2), dpi=100) #pass in tuple for the figure size, (width, height) in inches

ax = fig.add_axes([0,0,1,1])
ax.plot(x,x**2, label='X Squared', color='red', linewidth=3, alpha=0.5, linestyle='dashed', marker='o')
ax.plot(x,x**3, label='X Cubed', color='blue')

#Location code coincides with matplotlib location string. loc=0 sets legend to the "Best" spot, loc=10 is center, etc.
ax.legend(loc=0) 
#ax.legend(loc=(0.1,0.1)) #Sets location manually, 0 to 1

ax.set_title('Title')
ax.set_ylabel('Y Label')
ax.set_xlabel('X Label')
```

### Plot Limits

```python
fig = plt.figure(figsize=(4,2), dpi=100) #pass in tuple for the figure size, (width, height) in inches

ax = fig.add_axes([0,0,1,1])
ax.plot(x,x**2, label='X Squared', color='red')
ax.plot(x,x**3, label='X Cubed', color='blue')


#Location code coincides with matplotlib location string. loc=0 sets legend to the "Best" spot, loc=10 is center, etc.
ax.legend(loc=0) 
#ax.legend(loc=(0.1,0.1)) #Sets location manually, 0 to 1

ax.set_title('Title')
ax.set_ylabel('Y Label')
ax.set_ylim([0,1])
ax.set_xlabel('X Label')
ax.set_xlim([0,1])
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(8,2))

axes[0].plot(x,y)

axes[1].plot(y,x)

plt.tight_layout()
```
