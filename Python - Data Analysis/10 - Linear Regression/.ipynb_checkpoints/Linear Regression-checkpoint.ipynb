# Linear Regression

• Regression is predicting continuous values  
• Accuracy and Recall are NOT useful for calculating performance of a regression test. We instead would use the following metrics:

• Mean Absolute Error (MAE)  
-- Compare your predictions to the true Y-label  
-- Predicted house price vs. True house price  
-- WON'T punish large errors

• Mean Squared Error (MSE)  
-- Mean of the squared error  
-- Larger errors are noted more than with MAE, making this the more popular metric than MA  
-- ISSUE: Our units are also squared, meaning if we are calculating home prices, then that gets squared too ($100,000 becomes squared) and difficult to interpret, solve this by using RMSE

• Root Mean Square Error (RMSE)  
-- The square root of the MSE  
-- Most popular because it punishes larger error values, and has the same units as Y  
-- "Is this value good?": Context is everything, RMSE of 10 dollars for a house is good, whereas 10 dollars RMSE for a candy bar, is really bad  
-- Compare your metric to the average value of the label in your data set to try to get an intuition of its overall performance  
-- Domain knowledge is important!


# Theory

* The goal is to draw a line as close to each data point as we can get <br>
* For classic linear regression (Least Squares Method) you only measure the closeness in the "up and down" direction
* Least Squares Method is fitted by minimizing the sum of squares of the residuals (the difference between the observed value (y-value) and the fitted line)
