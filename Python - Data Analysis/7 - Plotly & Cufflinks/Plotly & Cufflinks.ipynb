# Plotly & Cufflinks

```python
import pandas as pd
import numpy as np
import cufflinks as cf
from plotly.offline import download_plotlyjs, init_notebook_mode, plot, iplot
%matplotlib inline
```

```python
from plotly import __version__
print(__version__)
```

```python
#Allows notebook to access visualizations
init_notebook_mode(connected=True)
```

```python
#Allows us to use cufflinks offline
cf.go_offline()
```

```python
import warnings
warnings.filterwarnings("ignore") # To ignore all warnings
```

```python
df = pd.DataFrame(np.random.randn(100,4), columns='A B C D'.split())
df
```

```python
#Normal line plot
df.plot()
```

### iplot

* Click and drag to zoom into a section of the plot
 * Double-click to zoom out

```python
df.iplot()
```

##### Scatter

```python
#Plotly tries to initially create a line between all points, add mode='markers' to create individual points
df.iplot(kind='scatter', x='A', y='B', mode='markers', size=15)
```

```python
#Columns need to be numerical
df.scatter_matrix()
```

##### Box

```python
df.iplot(kind='box')
```

##### Histogram

```python
df['A'].iplot(kind='hist', bins=25)
```

##### Spread

```python
df[['A', 'B']].iplot(kind='spread')
```

##### Bubble

```python
df.iplot(kind='bubble', x='A', y='B', size='C')
```

```python
df2 = pd.DataFrame({'Category': ['A','B', 'C'], 'Values': [32,43,50]})
df2
```

##### Bar

```python
df2.iplot(kind='bar', x='Category', y='Values')
```

##### 3D Surface Plot

```python
df3 = pd.DataFrame({'x': [1,2,3,4,5], 'y':[10,20,30,20,10], 'z': [500,400,300,200,100]})
df3
```

```python
df3.iplot(kind='surface', colorscale='blues')
```
