# Big Data - Hadoop, MapReduce, Spark, & PySpark

### Theory
Distributed Systems
* Can have one machine controlling the distribution of multiple machines
* Evens the computing power load across multiple machines, allowing us to use more processing power than a local machine
* Can easily add more machines as opposed to local machines
* Include fault tolerance, if one machine fails, the whole network can still go on

Hadoop
* A way to distribute very large files across multiple machines
* Uses Hadoop Distributed File System (HDFS)
* HDFS allows a user to work with large data sets
* HDFS also duplicates blocks of data for fault tolerance
* Then uses MapReduce which allows computations on the data
* HDFS uses a block of data (usually 128 MB by default)
  * If one machine gets knocked out, we still have our block of data on the other machines without losing any information
  * Smaller blocks provide more parallelization during processing

MapReduce
* A way of splitting a computation task to a distributed set of files
* Consists of a job Tracker and multiple Task Trackers
* Job Tracker sends code to run on the Task Trackers
* Task Trackers allocate CPU and memory for the tasks and monitor the tasks on the worker nodes

Spark
* Quickly and handle big data
* Can run on top of existing Hadoop HDFS systems
* Can use data stored in a variety of formats
  * Cassandra
  * AWS S3
  * HDFS & more
* MapReduce requires files to be stored in HDFS and Spark does not
* Can also perform operations up to 100x faster than MapReduce
  * MapReduce writes most of data to disk after each map and reduce operation
  * Spark keeps most of the data in memory after each transformation
  * Spark can spill over to disk if the memory is filled
* Resilient Distributed Dataset
  * Distributed Collection of Data
  * Fault-tolerant
  * Parallel operation - partitioned
  * Ability to use many data sources
  * RDD is also:
    * Immutable
    * Lazily Evaluated
    * Cacheable
  * Two types of RDD operations:
    * Transformations
      * Filter: RDD.filter(), applies a function to each elements and returns elements that evaluate to true
      * Map: RDD.map(), transforms each element and preserves the same # of elements, similar to pandas apply()
      * Flatmap: RDD.flatMap(), transforms each element into 0-N elements and changes # of elements
    * Actions
      * First: Return the first element in the RDD
      * Collect: Return all the elements of the RDD as an array at the driver program
      * Count: Return the number of elements in the RDD
      * Take: Return an array with the first n elements of the RDD
  * Usually hold values in tuples. Offers better partitioning of data
  * Reduce()
    * An action that will aggregate RDD elements using a function that returns a single element
  * ReduceByKey()
    * An action that will aggregate Pair RDD elements using a function that returns a Pair RDD
  * Reduce() and ReduceByKey() are similar to a group by operation

EC2 (Elastic Compute Cloud)
* A web service in AWS that provides resizable compute capacity in the cloud
* Think of it as a virtual computer we can access through the internet
* THE PLAN:
  * Create EC2 instance on AWS
  * Use SSH to connect to EC2 over the internet
  * SSH is different for Windows vs Mac/Linux
  * Set-up Spark and Jupyter on EC2 instance
