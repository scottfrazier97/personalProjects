# Big Data - Hadoop, MapReduce, Spark, & PySpark

### Theory
Distributed Systems
* Can have one machine controlling the distribution of multiple machines
* Evens the computing power load across multiple machines, allowing us to use more processing power than a local machine
* Can easily add more machines as opposed to local machines
* Include fault tolerance, if one machine fails, the whole network can still go on

Hadoop
* A way to distribute very large files across multiple machines
* Uses Hadoop Distributed File System (HDFS)
* HDFS allows a user to work with large data sets
* HDFS also duplicates blocks of data for fault tolerance
* Then uses MapReduce which allows computations on the data
* HDFS uses a block of data (usually 128 MB by default)
  * If one machine gets knocked out, we still have our block of data on the other machines without losing any information
  * Smaller blocks provide more parallelization during processing

MapReduce
* A way of splitting a computation task to a distributed set of files
* Consists of a job Tracker and multiple Task Trackers
* Job Tracker sends code to run on the Task Trackers
* Task Trackers allocate CPU and memory for the tasks and monitor the tasks on the worker nodes
